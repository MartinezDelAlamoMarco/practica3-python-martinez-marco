## Ejercicio 1 - Catálogo de productos como lista

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

print("Elementos en la lista:", len(catalogo))
print("Productos en el catálogo:", len(catalogo) // 2)

# los nombres van en las posiciones pares y los precios en las impares
nombres = catalogo[0::2]
precios = catalogo[1::2]
print(nombres)
print(precios)

# busco la posición del nombre y el precio está justo detrás
pos_nombre = catalogo.index("Robot Aspirador")
print("Precio del Robot Aspirador:", catalogo[pos_nombre + 1])

ultimos_tres_nombres = nombres[-3:]
print(ultimos_tres_nombres)

precio_medio = round(sum(precios) / len(precios), 2)
print("Precio medio:", precio_medio)
```

## Ejercicio 2 - Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

print("Producto de la 4ª fila:", inventario[3][1])

valor_stock = []
for fila in inventario:
    producto = fila[1]
    valor_fila = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor_fila])
print(valor_stock)

valor_total = round(sum(linea[1] for linea in valor_stock), 2)
print("Valor inmovilizado total:", valor_total)

# el ALM-SUR ocupa las filas 2 y 3, así que van seguidas
filas_alm_sur = inventario[2:4]
valor_alm_sur = round(sum(fila[2] * fila[3] for fila in filas_alm_sur), 2)
print("Valor ALM-SUR:", valor_alm_sur)
```

El slicing `inventario[2:4]` funciona porque sé que esas dos filas del ALM-SUR están juntas, pero si se añade una fila nueva o los almacenes se desordenan, ese mismo slice me daría datos de otro almacén sin avisar de nada. Por eso no es fiable filtrar por posición: hay que filtrar comparando el valor (`fila[0] == "ALM-SUR"`), que no depende de dónde esté cada fila.

## Ejercicio 3 - Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

# esto no copia la lista, apunta al mismo objeto
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print(precios_rebajados)
print(precios_originales)

precios_originales[0] = 59.90
copia_con_list = list(precios_originales)
copia_con_slice = precios_originales[:]

copia_con_list[0] = 0.01
copia_con_slice[1] = 0.02
print(precios_originales)
print(copia_con_list)
print(copia_con_slice)

rebajada = precios_originales[:]
for i in range(len(rebajada)):
    rebajada[i] = round(rebajada[i] * 0.90, 2)
print(rebajada)

rebajada.extend([99.90, 45.00])
rebajada.insert(2, 129.90)
print(rebajada)

del rebajada[4]
descartado = rebajada.pop()
print(descartado, rebajada)

print(id(precios_originales), id(copia_con_list), id(copia_con_slice))
print(precios_originales is copia_con_list)
```

## Ejercicio 4 - Cola de pedidos pendientes

```python
cola = ["PED-10021", "PED-10007", "PED-10044", "PED-10012", "PED-10033", "PED-10008"]

cola_invertida = cola[::-1]
print(cola_invertida)

cola_ordenada = sorted(cola)
print(cola_ordenada)
print(cola)

cola.sort(reverse=True)
print(cola)

atendidos = [cola.pop(0), cola.pop(0)]
print(atendidos)

cola.insert(0, "PED-99999")
print(cola)

total_pedidos = len(atendidos) + len(cola)
avance_pct = round(len(atendidos) / total_pedidos * 100, 1)
print(f"Atendidos: {len(atendidos)} | Pendientes: {len(cola)} | Avance: {avance_pct} %")
```
