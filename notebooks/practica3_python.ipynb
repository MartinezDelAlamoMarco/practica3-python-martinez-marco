## Ejercicio 2 - Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

print("Producto de la 4ª fila:", inventario[3][1])

valor_stock = []
for fila in inventario:
    producto = fila[1]
    valor_fila = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor_fila])
print(valor_stock)

valor_total = round(sum(linea[1] for linea in valor_stock), 2)
print("Valor inmovilizado total:", valor_total)

# el ALM-SUR ocupa las filas 2 y 3, así que van seguidas
filas_alm_sur = inventario[2:4]
valor_alm_sur = round(sum(fila[2] * fila[3] for fila in filas_alm_sur), 2)
print("Valor ALM-SUR:", valor_alm_sur)
```

El slicing `inventario[2:4]` funciona porque sé que esas dos filas del ALM-SUR están juntas, pero si se añade una fila nueva o los almacenes se desordenan, ese mismo slice me daría datos de otro almacén sin avisar de nada. Por eso no es fiable filtrar por posición: hay que filtrar comparando el valor (`fila[0] == "ALM-SUR"`), que no depende de dónde esté cada fila.

## Ejercicio 5 - Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Importe neto de una línea de pedido, con descuento opcional."""
    bruto = unidades * precio_unitario
    return round(bruto * (1 - descuento_pct / 100), 2)


def detalle_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve (importe_bruto, descuento_aplicado, importe_neto)."""
    importe_bruto = round(unidades * precio_unitario, 2)
    importe_neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    descuento_aplicado = round(importe_bruto - importe_neto, 2)
    return importe_bruto, descuento_aplicado, importe_neto


importe_bruto, descuento_aplicado, importe_neto = detalle_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print(f"Bruto: {importe_bruto} | Descuento: {descuento_aplicado} | Neto: {importe_neto}")

help(calcular_importe)

lista_pedidos = [(3, 59.90, 10), (1, 279.00, 0), (5, 18.50, 20), (2, 149.00, 5), (4, 89.90, 15)]
importes_calculados = [calcular_importe(u, p, d) for u, p, d in lista_pedidos]

print(importes_calculados)
print("Máximo:", max(importes_calculados))
print("Mínimo:", min(importes_calculados))
print("Ordenados:", sorted(importes_calculados))
print("Nº pedidos:", len(importes_calculados), "| Total:", round(sum(importes_calculados), 2))
```

## Ejercicio 8 - Del cálculo con listas al cálculo vectorizado

```python
unidades_pedido = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios_pedido = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

try:
    resultado = unidades_pedido * precios_pedido
except TypeError as e:
    print("TypeError:", e)
```

Con listas de Python, `*` significa repetir la secuencia, no multiplicar elemento a elemento, así que solo acepta un entero a la derecha; por eso multiplicar dos listas revienta con `TypeError`.

```python
arr_unidades = np.array(unidades_pedido)
arr_precios = np.array(precios_pedido)
print(arr_unidades.dtype, arr_unidades.shape)
print(arr_precios.dtype, arr_precios.shape)

importes = arr_unidades * arr_precios
print(importes)
print("Total:", round(importes.sum(), 2))

importes_con_recargo = importes * 1.04
print(np.round(importes_con_recargo, 2))
print("Total con recargo:", round(importes_con_recargo.sum(), 2))

array_mixto = np.array([1, "dos", 3.0, True])
print(array_mixto, array_mixto.dtype)
```

El array mixto sale como cadenas (`<U32`) porque un `ndarray` guarda todos sus elementos en memoria con el mismo tipo, y NumPy convierte todo al tipo más general capaz de representarlo, que aquí es texto.

## Ejercicio 11 - Estadística descriptiva sobre telemetría simulada

```python
generador = np.random.default_rng(2024)
temperaturas = generador.normal(68, 6.5, 500)
vibraciones = np.abs(generador.normal(2.4, 0.8, 500))

media_temp = temperaturas.mean()
mediana_temp = np.median(temperaturas)
desv_temp = temperaturas.std()
print(f"Media: {media_temp:.2f} | Mediana: {mediana_temp:.2f} | Desv: {desv_temp:.2f}")
print(f"Mín: {temperaturas.min():.2f} | Máx: {temperaturas.max():.2f}")

percentiles_vibracion = np.percentile(vibraciones, [25, 50, 75, 95])
print(np.round(percentiles_vibracion, 3))

matriz_datos = np.column_stack((temperaturas, vibraciones))
print(matriz_datos.shape)

correlacion = np.corrcoef(temperaturas, vibraciones)
print(np.round(correlacion, 4))

limite_temp = media_temp + 2 * desv_temp
lecturas_altas = (temperaturas > limite_temp).sum()
print(f"Umbral: {limite_temp:.2f} | Lecturas: {lecturas_altas} | {lecturas_altas/500*100:.2f} %")
```

Media y mediana salen casi iguales, así que la distribución es simétrica y no hay valores raros tirando de la media hacia un lado.

La correlación sale prácticamente 0, así que temperatura y vibración no están relacionadas linealmente en estos datos (tiene sentido, las generé por separado).

Fijar la semilla (`default_rng(2024)`) hace que la secuencia de números aleatorios sea siempre la misma, así que si el profesor ejecuta esta celda le van a salir exactamente los mismos valores que a mí y puede comprobar el resultado.

## Ejercicio 15 - Columnas calculadas, agregación y exportación

```python
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]
ventas["importe_neto"] = ventas["importe_bruto"] * (1 - ventas["descuento_pct"] / 100)
ventas["ticket_medio"] = ventas["importe_neto"] / ventas["unidades"]

print(ventas[["unidades", "precio_unitario", "descuento_pct",
              "importe_bruto", "importe_neto", "ticket_medio"]].head().round(2))
```

```python
facturacion_por_zona = ventas.groupby("region")["importe_neto"].sum().sort_values(ascending=False)
print(facturacion_por_zona.round(2))
print("Facturación total:", round(ventas["importe_neto"].sum(), 2))
```

```python
resumen_cat_canal = ventas.groupby(["categoria", "canal"]).agg({
    "importe_neto": "mean",
    "unidades": "sum",
}).round(2)
print(resumen_cat_canal)
```

```python
top5_productos = ventas.groupby("producto")["importe_neto"].sum().nlargest(5)
print(top5_productos.round(2))
```

```python
mediana_satisfaccion = ventas["satisfaccion"].median()
print("Mediana satisfacción:", mediana_satisfaccion)

ventas["satisfaccion"] = ventas["satisfaccion"].fillna(mediana_satisfaccion)
ventas["canal"] = ventas["canal"].fillna("Desconocido")
print(ventas[["satisfaccion", "canal"]].isna().sum())
print(ventas["canal"].value_counts())

facturacion_por_zona.round(2).to_csv("../outputs/facturacion_por_region.csv", header=["importe_neto"])
print("Exportado a ../outputs/facturacion_por_region.csv")
```
